#### **NORMALIZACIÓN**

En ocasiones, ciertos registros de texto suelen tener **símbolos especiales** como `!!!`, `@@@`, espacios extra, etc. Con el uso de **Expresiones Regulares (REGEX) se limpia el texto de los símbolos expeciales.**

````python
# Observar valores únicos por columna de texto
df_yt['categoría'].unique()
````

````python
df_yt['título'].unique()
````

````python
# Definir función de eliminar caracteres especiales y espacios extras
import re

def limpiar_texto(texto):
  if pd.isna(texto): return texto
  texto = re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ0-9\s]', '',str(texto))
  return texto.strip().lower()
````

````python
# Aplicar limpieza a columnas con texto
df_yt['categoría'] = df_yt['categoría'].apply(limpiar_texto)
df_yt['título'] = df_yt['título'].apply(limpiar_texto)
````

````python
# Observar cambios en los valores únicos
df_yt['categoría'].unique()
````

````python
df_yt['título'].unique()
````

````python
# Corregir detalles de texto adicionales
reemplazos = {
    'musica': 'música',
    'musíca': 'música',
    'musicá': 'música',
    'tecnologia': 'tecnología',
    'educacion': 'educación',
    'peliculas': 'películas'
}
df_yt['categoría'] = df_yt['categoría'].replace(reemplazos)
df_yt['categoría'].unique()  # Se observa catálogo de datos
````

````python
df_yt['categoría']
````

````python
# Cambio de tipo de dato

# Los catálogos de datos deben ser del tipo de dato category
df_yt['título'] = df_yt['título'].astype('category')
df_yt['categoría'] = df_yt['categoría'].astype('category')

print(df_yt['título'].dtype)
print(df_yt['categoría'].dtype)
````

````python
# Verificar cambio de típos de dato
df_yt.info()
````

#### **IMPUTACIÓN Y NORMALIZACIÓN**

En la fase de **imputación y normalización para variables numéricas**, el objetivo es asegurar que estos datos sean consistentes, completos y adecuados para el análisis.

Estas acciones son cruciales para mantener la integridad del dataset y garantizar que los análisis posteriores no se vean afectados por datos faltantes o erróneos.

````python
col_num_null = ['views', 'likes']
df_yt[col_num_null].describe()
````

````python
# Observar distribución de las columnas con nulos

columnas_a_graficar = ['views', 'likes']
plt.figure(figsize=(15, 6))

for i, col in enumerate(columnas_a_graficar, 1):
    plt.subplot(1, 2, i)
    sns.histplot(df_yt[col], bins=30, kde=True, color='teal')
    plt.title(f'Distribución de {col.replace("_", " ").title()}')
    plt.xlabel(col.replace("_", " ").title())
    plt.ylabel('Frecuencia')
    plt.grid(axis='y', alpha=0.75)

plt.tight_layout()
plt.show()

# Se observan distribuciones con valores extremos (no normales)
````

````python
# Calcular media y desviación estándar de las columnas con nulos
# Se busca demostrar visualmente la lejanía de los datos hacia la media
media_views = df_yt['views'].mean()
median_views = df_yt['views'].median()
std_views = df_yt['views'].std()

media_likes = df_yt['likes'].mean()
median_likes = df_yt['likes'].median()
std_likes = df_yt['likes'].std()

lim_sup_views = media_views + std_views
lim_inf_views = media_views - std_views

lim_sup_likes = media_likes + std_likes
lim_inf_likes = media_likes - std_likes

configuracion_views = [(media_views, 'blue'), (lim_sup_views, 'red'), (lim_inf_views, 'red'), (median_views, 'green')]
configuracion_likes = [(media_likes, 'blue'), (lim_sup_likes, 'red'), (lim_inf_likes, 'red'), (median_likes, 'green')]
````

````python
# Observar distribución de las columnas con nulos

plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1) # Los datos se alejan de la media
sns.scatterplot(x=range(len(df_yt)), y=df_yt['views'], s=10, color='gray')
plt.title('Desviación Estandar de Vistas')
plt.xlabel('Índice del registro')
plt.ylabel('Vistas')
for dato, color in configuracion_views:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.subplot(1, 2, 2) # Los datos se alejan de la media y se polarizan en dos extremos
sns.scatterplot(x=range(len(df_yt)), y=df_yt['likes'], s=10, color='gray')
plt.title('Desviación Estandar de Likes')
plt.xlabel('Índice del registro')
plt.ylabel('Likes')
for dato, color in configuracion_likes:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.tight_layout()
plt.show()
````

## **EXPLORACIÓN DE DATOS**

````python
# Imputación de datos

# Debido a los outliers y la alejanía hacia la media, la MTC será la mediana
df_yt['views'] = df_yt['views'].fillna(df_yt['views'].median())
df_yt['likes'] = df_yt['likes'].fillna(df_yt['likes'].median())

df_yt.loc[df_yt['views'] < 0, 'views'] = df_yt['views'].median() # No existen vistas negativas, imputar con la mediana
print(df_yt.isnull().sum())
````

````python
# Cambio de tipo de dato

# La naturaleza de las vistas y likes es ser enteros
df_yt['views'] = df_yt['views'].astype('int')
df_yt['likes'] = df_yt['likes'].astype('int')
````
